```python
import statistics
scores = [56,85,64,98,65,46,75,37,84,68]
mean   = statistics.mean(scores)
median = statistics.median(scores)
mode   = statistics.mode(scores)
print(f'Data   : {sorted(scores)}')
print(f'Mean   : {mean:.2f}')
print(f'Median : {median}')
print(f'Mode   : {mode}')
```

```python
data_range  = max(scores) - min(scores)
variance    = statistics.variance(scores)
std_dev     = statistics.stdev(scores)
print(f'Range      : {data_range}')
print(f'Variance   : {variance:.2f}')
print(f'Std Dev    : {std_dev:.2f}')
```

```python
data = [3, 5, 5, 6, 7, 8, 9, 11]

n = len(data)

mean_val = sum(data) / n
print("Mean =", mean_val)

deviations = [x - mean_val for x in data]
squared_devs = [d ** 2 for d in deviations]

variance = sum(squared_devs) / n
std_dev = variance ** 0.5

print("Deviations :", deviations)
print("Squared deviations :", squared_devs)
print("Variance =", round(variance, 2))
print("Standard deviation =", round(std_dev, 2))
```

```python
total = 8

p_five = 1 / total
print("P(getting 5) =", round(p_five, 4))

even = 4
p_even = even / total
print("P(even number) =", round(p_even, 4))

p_odd = 1 - p_even
print("P(odd number) =", round(p_odd, 4))

less_than_4 = 3
p_less_4 = less_than_4 / total
print("P(number less than 4) =", round(p_less_4, 4))

```

```python
import random

random.seed(25)
n_rolls = 5000

count_even = 0

for _ in range(n_rolls):
    roll = random.randint(1, 6)
    if roll % 2 == 0:
        count_even += 1

experimental_probability = count_even / n_rolls

print("Rolls =", n_rolls)
print("Even count =", count_even)
print("Experimental P(even) =", round(experimental_probability, 4))
print("Theoretical P(even) = 0.5000")
```
